# Starter EDA (work/01_eda_starter)

This notebook runs quick checks on the starter CSV so you can iterate fast. It is the same code used in the Colab demo you ran. Run this notebook in Colab or Jupyter.

Sections:
- Load CSV
- Quick checks: shape, missingness
- Apply gotchas: avg_position=0 -> NaN; create has_* flags
- Simple plots
- Save small cache to work/outputs/eda_starter.parquet


```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
from pathlib import Path
Path('work/outputs').mkdir(parents=True, exist_ok=True)
```

```python
# Load starter CSV from repo (or upload locally in Colab)
url = 'https://raw.githubusercontent.com/Noorulhuda-12/flyrank/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url, low_memory=False)
print('Loaded shape:', df.shape)
```

```python
# Quick checks
print('Rows, cols:', df.shape)
print('Top missing columns (percent):')
print((df.isna().mean() * 100).round(2).sort_values(ascending=False).head(10))
```

```python
# Apply gotchas: avg_position = 0 means missing
if 'avg_position' in df.columns:
    df.loc[df['avg_position']==0, 'avg_position'] = np.nan
    print('Converted avg_position 0 to NaN')
```

```python
# Create presence flags
df['has_keywords'] = df['keywords'].notna().astype(int) if 'keywords' in df.columns else 0
df['has_word_count'] = df['word_count'].notna().astype(int) if 'word_count' in df.columns else 0
print('Created has_* flags')
```

```python
# Simple plots
import matplotlib.pyplot as plt
if 'ctr' in df.columns:
    plt.figure(figsize=(8,3))
    df['ctr'].dropna().hist(bins=40)
    plt.title('CTR distribution')
    plt.show()
```

```python
# Save a small cache for later notebooks
cache_cols = ['content_id','client_id','ctr','avg_position','content_type','word_count']
cache_cols = [c for c in cache_cols if c in df.columns]
df[cache_cols].to_parquet('work/outputs/eda_starter.parquet', index=False)
print('Saved cache to work/outputs/eda_starter.parquet')
```
