## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# Section 3: Verification Queries
# Validating the data contract claims made in Sections 1 and 2

import pandas as pd
import numpy as np

# Creating a dummy dataset to simulate verification
# In a real scenario, this would be your BigQuery or CSV source
data = {
    'session_id': [1, 2, 2, 3, 4, 5],
    'search_query': ['python', 'data contract', 'data contract', 'mlops', 'sql', None],
    'click_binary': [1, 0, 0, 1, 0, 0],
    'timestamp': pd.to_datetime(['2023-01-05', '2023-01-10', '2023-01-10', '2023-02-15', '2023-03-20', '2023-03-25'])
}
df = pd.DataFrame(data)

print(f"--- Verification for Skill: {required_skill} ---")

# 1. Grain Check: Is 'one row = one search session'?
duplicates = df.duplicated().sum()
unique_sessions = df['session_id'].nunique()
total_rows = len(df)
print(f"Grain Check: Found {unique_sessions} unique sessions across {total_rows} rows.")
print(f"Duplicates found: {duplicates}")

# 2. Missing Values Check
missing_features = df['search_query'].isnull().sum()
print(f"Missing Values in 'search_query': {missing_features}")

# 3. Time Window Check
min_date = df['timestamp'].min()
max_date = df['timestamp'].max()
print(f"Observed Time Window: {min_date.date()} to {max_date.date()}")
print(f"Matches Contract ({start_date} to {end_date}): {min_date >= pd.to_datetime(start_date) and max_date <= pd.to_datetime(end_date)}")
```
